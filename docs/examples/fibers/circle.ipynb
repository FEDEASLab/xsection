# Fibers: Circle Layouts

```python
import veux
import xara
from xsection.library import Circle
```

```python
circle = Circle(radius=5)

veux.draw_shape(circle)
```

```python
fibers = {"d": 1}
section = xara.FrameSection("UniaxialFiber", circle, fibers=fibers)
veux.draw_shape(section)
```

```python
fibers = {"r": 4, "rule": "mid"}
section = xara.FrameSection("UniaxialFiber", circle, fibers=fibers)
veux.draw_shape(section)
```

```python
fibers = {"r": 4, "rule": "sunflower"}
section = xara.FrameSection("UniaxialFiber", circle, fibers=fibers)
veux.draw_shape(section)
```
