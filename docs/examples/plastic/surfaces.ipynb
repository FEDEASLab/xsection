```python
from xsection.library import from_aisc 

shape = from_aisc("W18x40", mesh_scale=1)
```

```python
import veux
veux.draw_shape(shape)
```

```python
from xsection.analysis.plastic_analysis import PlasticAnalysis
import matplotlib.pyplot as plt

analysis = PlasticAnalysis(shape, Fy=60)
plt.plot(*analysis.plastic_contour([0,1,0]))

plt.ylabel("N")
plt.xlabel("M")
```
