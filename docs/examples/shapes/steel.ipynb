# Steel Shapes

```python
import veux

import xara
material = xara.MultiaxialMaterial("ElasticIsotropic", E=1, G=1)
```

## AISC

The [`from_aisc`](https://peer-open-source.github.io/xsection/api/library/from_aisc.html) function generates standard steel shapes using an AISC identifier. 
For example, the following code creates a [`WideFlange`](https://peer-open-source.github.io/xsection/api/library/WideFlange.html) object with the dimensions of an AISC `W14x211` shape.

```python
from xsection.library import from_aisc, load_shape

shape = from_aisc("W14x211", fillet=True)
veux.draw_shape(shape)
```

```python
print(shape.elastic.Iy)
```

The `summary()` method can also be used to print all properties at once:

```python
print(shape.elastic.summary())
```

### Mesh Size

The size of the mesh is controlled by the `mesh_scale` argument. 

```python
shape = from_aisc("W14x211", fillet=True, mesh_scale=1/2)
veux.draw_shape(shape)
```

```python
shape = from_aisc("W14x211", fillet=True, units="mks")
print(shape.elastic.summary())
```

## HSS

```python
shape = load_shape("HSS5x5x3/8", library="AISC16", 
                   material=material,
                   mesh_scale=1/3, mesher="gmsh")
print(shape.elastic.summary())
print(xara.FrameSection("Elastic", shape).constants)
veux.draw_shape(shape)
```
